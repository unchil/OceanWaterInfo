```kotlin
val df_day_raw = DataFrame.readJson("http://192.168.35.107:7788/nifs/seawaterinfo/oneday").parse()

df_day_raw.describe()
```

```kotlin
df_day_raw.plot {

    layout {
        title = "일별 해수 온도 통계"
        size = 2000 to 800
        x.axis.name = "Observatory"
        y.axis.name = "Temperature"
    }

    boxplot(sta_nam_kor, wtr_tmp) {
        boxes {
            fillColor(Stat.middle) {
                legend{
                    name = "Temperature "
                    breaks(format = "{d}°C")
                }
                scale = continuous(Color.BLUE..Color.RED)
            }
            // Border line color depends on `x` category
            borderLine.color(Stat.x)
            // `identity` position, i.e boxes are overlapping
        }

        outliers {
            // points context + StatBoxplotOutliers context
            // color depends on `x` category
            color(Stat.x){
                legend.name = "Observatory"
            }
        }
    }
}
```

```kotlin
var df_current_raw = DataFrame.readJson("http://192.168.35.107:7788/nifs/seawaterinfo/current").parse()

df_current_raw.describe()
```

```kotlin
val df_current =  df_current_raw.insert("관측층"){
    when(this.obs_lay){
        1 -> "표층"
        2 -> "중층"
        3 -> "심층"
        else -> ""
    }
}.after("obs_lay")

df_current.describe()
```

```kotlin
df_current
    .select{sta_nam_kor and wtr_tmp and "관측층"}
    .plot {
        layout {
            title = "관측 층별 해수 온도"
            size = 2600 to 600
            xAxisLabel = "Observatory"
            yAxisLabel = "Temperature( °C )"
        }

        bars {
            x(sta_nam_kor)

            y(wtr_tmp)

            fillColor("관측층") {
                legend{
                    name = "Ocean Layer"
                }
                scale =  categoricalColorHue()
            }
        }
    }
```
